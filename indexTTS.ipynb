```python
import subprocess
import threading
import time
import os
import requests

# 1. 自动校准路径
if os.path.exists("/content/index-tts/index-tts"):
    %cd /content/index-tts/index-tts
elif os.path.exists("/content/index-tts"):
    %cd /content/index-tts
else:
    print("❌ 错误：找不到项目目录，请确保你已经运行了前面的 Git Clone 单元格")

# 2. 检查并下载 cloudflared (防止文件丢失)
if not os.path.exists("cloudflared"):
    print("正在下载 cloudflared...")
    !wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O cloudflared
    !chmod +x cloudflared

def start_webui():
    print("正在后台加载模型 (约 1 分钟)...")
    # 尝试使用 uv 运行，如果报错则回退到系统 python
    try:
        subprocess.run("uv run python webui.py --port 7860", shell=True)
    except:
        subprocess.run("python webui.py --port 7860", shell=True)

# 3. 启动 WebUI 线程
threading.Thread(target=start_webui, daemon=True).start()

# 4. 循环检查 WebUI 是否真的起来了
print("正在等待端口 7860 响应...")
max_retries = 20
for i in range(max_retries):
    time.sleep(10)
    try:
        # 尝试访问本地端口
        requests.get("http://127.0.0.1:7860")
        print("\n✅ WebUI 已就绪！")
        break
    except:
        print(f"  等待中 ({i+1}/{max_retries})...")

# 5. 启动穿透
print("="*50)
print("🚀 点击下方出现的 trycloudflare 链接即可访问")
print("="*50)
!./cloudflared tunnel --url http://127.0.0.1:7860
```
